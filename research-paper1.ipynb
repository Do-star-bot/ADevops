```python
import os
import pandas as pd

# Find CSV files
csv_files = []

for root, dirs, files in os.walk("/kaggle/input"):
    for file in files:
        if file.endswith(".csv"):
            csv_files.append(os.path.join(root, file))

if not csv_files:
    print("No CSV files found. Check whether the dataset is attached.")
else:
    print("Available CSV files:")
    for i, path in enumerate(csv_files):
        print(f"{i}: {path}")

    # Load the first CSV file
    df = pd.read_csv(csv_files[0])

    print("\nDataset shape:", df.shape)
    print("\nColumn names:")
    print(df.columns.tolist())

    print("\nFirst 5 rows:")
    display(df.head())
```

```python
# Check dataset information
print("Dataset shape:", df.shape)

print("\nColumn names and data types:")
df.info()

print("\nMissing values in each column:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nNumber of duplicate rows:", df.duplicated().sum())
```

```python
# Display sample customer feedback from each text column
text_columns = df.select_dtypes(include=["object", "string"]).columns

for col in text_columns:
    print(f"\n{'=' * 60}")
    print(f"Column: {col}")
    print(df[col].dropna().astype(str).head(3).to_list())
```

```python
import pandas as pd

# Select the correct customer feedback column
TEXT_COLUMN = "Ticket_Description"

# Remove missing and empty feedback
df_clean = df.dropna(subset=[TEXT_COLUMN]).copy()

df_clean[TEXT_COLUMN] = df_clean[TEXT_COLUMN].astype(str).str.strip()
df_clean = df_clean[df_clean[TEXT_COLUMN] != ""]

# Remove duplicate feedback
df_clean = df_clean.drop_duplicates(subset=[TEXT_COLUMN]).reset_index(drop=True)

print("Original dataset size:", len(df))
print("Clean dataset size:", len(df_clean))
print("Feedback column:", TEXT_COLUMN)

print("\nSample customer feedback:")
display(df_clean[[TEXT_COLUMN, "Issue_Category", "Satisfaction_Score"]].head())
```

```python
# Select 20 records for initial LLM experiments
sample_size = min(20, len(df_clean))

df_sample = df_clean.sample(
    n=sample_size,
    random_state=42
).reset_index(drop=True)

# Create a sequential ID for each sampled ticket
df_sample["Sample_ID"] = range(1, len(df_sample) + 1)

print("Total cleaned records:", len(df_clean))
print("Records selected for LLM:", len(df_sample))

print("\nSample customer feedback:")
display(
    df_sample[
        ["Sample_ID", "Ticket_Description",
         "Issue_Category", "Satisfaction_Score"]
    ].head(5)
)
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

MODEL_NAME = "google/flan-t5-small"

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

# Load pretrained language model
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)

# Use GPU if available; otherwise use CPU
device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)

print("Model loaded successfully!")
print("Model:", MODEL_NAME)
print("Device:", device)
```

```python
import pandas as pd
import torch
from tqdm.auto import tqdm

def summarize_feedback(text):
    prompt = (
        "Summarize the following customer support feedback "
        "in one short sentence:\n" + str(text)
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        max_length=512,
        truncation=True
    ).to(device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=60,
            num_beams=4,
            do_sample=False
        )

    summary = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return summary

# Generate summaries
summaries = []

for text in tqdm(df_sample[TEXT_COLUMN], desc="Summarizing tickets"):
    summaries.append(summarize_feedback(text))

df_sample["LLM_Summary"] = summaries

print("Original feedback and generated summaries:")
display(
    df_sample[
        ["Sample_ID", TEXT_COLUMN, "LLM_Summary"]
    ].head(10)
)
```

```python
def extract_topic(text):
    prompt = (
        "Identify the single main issue discussed in this customer "
        "support ticket. Return only a short topic label of 2 to 5 words.\n\n"
        "Customer ticket: " + str(text) + "\n\n"
        "Topic:"
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        max_length=512,
        truncation=True
    ).to(device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=20,
            num_beams=4,
            do_sample=False
        )

    return tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    ).strip()


# Extract one topic per ticket
topics = []

for text in tqdm(df_sample[TEXT_COLUMN], desc="Extracting topics"):
    topics.append(extract_topic(text))

df_sample["LLM_Topic"] = topics

print("Extracted topics:")
display(
    df_sample[
        ["Sample_ID", TEXT_COLUMN, "LLM_Summary", "LLM_Topic"]
    ]
)
```

```python
def extract_topic(text):
    prompt = (
        "Identify the main issue in this customer support ticket. "
        "Return only a short topic label of 2 to 5 words.\n\n"
        "Customer ticket: " + str(text)
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        max_length=512,
        truncation=True
    ).to(device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=20,
            num_beams=4,
            do_sample=False
        )

    return tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    ).strip()


from tqdm.auto import tqdm

df_sample["LLM_Topic"] = [
    extract_topic(text)
    for text in tqdm(
        df_sample[TEXT_COLUMN],
        desc="Extracting topics"
    )
]

display(
    df_sample[
        ["Sample_ID", TEXT_COLUMN, "LLM_Topic"]
    ]
)
```

```python
import plotly.express as px
from textblob import TextBlob

# Calculate sentiment from customer feedback
def get_sentiment(text):
    polarity = TextBlob(str(text)).sentiment.polarity

    if polarity > 0.05:
        return "Positive"
    elif polarity < -0.05:
        return "Negative"
    else:
        return "Neutral"

df_clean["Sentiment"] = df_clean["Ticket_Description"].apply(get_sentiment)

# Count sentiment categories
sentiment_counts = (
    df_clean["Sentiment"]
    .value_counts()
    .rename_axis("Sentiment")
    .reset_index(name="Number of Tickets")
)

# Create interactive donut chart
fig = px.pie(
    sentiment_counts,
    names="Sentiment",
    values="Number of Tickets",
    title="Customer Feedback Sentiment Distribution",
    hole=0.45,
    color="Sentiment",
    color_discrete_map={
        "Positive": "#22C55E",
        "Neutral": "#64748B",
        "Negative": "#EF4444"
    },
    template="plotly_white"
)

fig.update_traces(
    textposition="inside",
    textinfo="label+percent",
    hovertemplate="%{label}<br>Tickets: %{value}<br>Share: %{percent}<extra></extra>"
)

fig.update_layout(title_x=0.5)
fig.show()

print("Sentiment counts:")
display(sentiment_counts)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import MiniBatchKMeans

# Prepare customer feedback
feedback = df_clean["Ticket_Description"].fillna("").astype(str)

# Convert feedback into TF-IDF features
vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=2000,
    ngram_range=(1, 2),
    min_df=2
)

tfidf_matrix = vectorizer.fit_transform(feedback)

print("TF-IDF matrix shape:", tfidf_matrix.shape)

# Discover 10 topics
kmeans = MiniBatchKMeans(
    n_clusters=10,
    random_state=42,
    n_init=10,
    batch_size=1024
)

df_clean["Topic_ID"] = kmeans.fit_predict(tfidf_matrix)

print("\nNumber of topics discovered:",
      df_clean["Topic_ID"].nunique())

print("\nFeedback count per topic:")
display(
    df_clean["Topic_ID"]
    .value_counts()
    .sort_index()
    .rename_axis("Topic_ID")
    .reset_index(name="Feedback_Count")
)
```

```python

import pandas as pd

# Get the TF-IDF feature names
feature_names = vectorizer.get_feature_names_out()

# Extract top keywords for each cluster
topic_keywords = {}

for topic_id in range(10):
    # Get the cluster center
    center = kmeans.cluster_centers_[topic_id]

    # Find the top 10 keywords
    top_indices = center.argsort()[-10:][::-1]
    keywords = feature_names[top_indices]

    topic_keywords[topic_id] = list(keywords)

# Display keywords for each topic
topic_table = pd.DataFrame({
    "Topic_ID": list(topic_keywords.keys()),
    "Top_Keywords": [
        ", ".join(keywords)
        for keywords in topic_keywords.values()
    ]
})

display(topic_table)
```
